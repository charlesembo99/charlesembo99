```python
#importing requared packages for the project with are pandas,requests, and Beatifulsoap4
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
#Inialiazing the headers and rows arrays variables for hockey titles headers and hockey table data respectively
headers = []
rows = []
```

```python
#Define the scrape_hockey_data function that accepts the number of pages you want to scrape
def scrape_hockey_data(num_pages):
  # Loop thruogh each page and add plus since index starts at 0
    for page in range(1,num_pages + 1):
        # set the base url for the site we want to scrape
        base_url = 'https://www.scrapethissite.com/pages/forms/'
        # Formart the url icluding each page we are on the loop
        url = base_url.format(page)
        # request the contents of the site
        response = requests.get(url)
        #parse the Html content
        soup = BeautifulSoup(response.text, 'html.parser')
        # access all the table element on the page
        hockey_table = soup.find('table', class_='table')
        #check if there is the table and return approapriate result or continue
        if not hockey_table:
            print(f"No table found on page {page}")
            continue
        # on our first iteration we are getting the headers or the hockey table titles and setting it the same for the subsequest page iteration
        if page == 1:
            headers= [th.get_text(strip=True) for th in hockey_table.find_all('th')]
        #Getting the rows and data in the rows for each row
        for tr in hockey_table.find_all('tr', class_='team'):
            #Data for each row
            cells = [td.get_text(strip=True) for td in tr.find_all('td')]
            if cells:
                #Adding to our row list which contains all of our data
                rows.append(cells)
    #Constructing a dataframe using rows which is our data and the columns
    df = pd.DataFrame(rows, columns=headers)
    return df
```

```python
#Calling the scrape_hockey , saving data to a csv file and completing the scraping
df = scrape_hockey_data(2)
df.to_csv(r'./Hockey.csv', index=False)
print("Scraping complete. Data saved to Hockey.csv.")
```
